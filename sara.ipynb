```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load telecom customer dataset
df = pd.read_csv("C:/Users/saara/Downloads/archive (3).zip")

# Convert MonthlyCharges to numeric
df['MonthlyCharges'] = pd.to_numeric(
    df['MonthlyCharges'], errors='coerce'
)

# Remove missing values
data = df['MonthlyCharges'].dropna()

# Plot histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data,
    bins=10,
    color="orange",
    edgecolor="black",
    kde=True
)

# Set title and labels
plt.title("Distribution of Telecom Customers")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")

plt.tight_layout()

# Display histogram
plt.show()

# Save histogram
plt.savefig("telecom_histogram.png", dpi=300)

print("Histogram generated successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read the telecom customer dataset
df = pd.read_csv("C:/Users/saara/Downloads/archive (3).zip")

# Convert MonthlyCharges to numeric
df['MonthlyCharges'] = pd.to_numeric(
    df['MonthlyCharges'], errors='coerce'
)

# Remove missing values
data = df['MonthlyCharges'].dropna()

# Create boxplot
plt.figure(figsize=(8, 5))

plt.boxplot(
    data,
    patch_artist=True,
    boxprops=dict(facecolor='orange', color='black'),
    medianprops=dict(color='red', linewidth=2),
    whiskerprops=dict(color='black'),
    capprops=dict(color='black'),
    flierprops=dict(marker='o', markerfacecolor='blue',
                    markersize=5)
)

# Add title and labels
plt.title("Boxplot of Telecom Customer Monthly Charges")
plt.xlabel("Telecom Customers")
plt.ylabel("Monthly Charges")

# Display boxplot
plt.tight_layout()
plt.show()

print("Telecom customer boxplot generated successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read telecom customer dataset
df = pd.read_csv("C:/Users/saara/Downloads/archive (3).zip")

# Count the top 10 customer categories
customer_counts = df['InternetService'].value_counts().head(10)

# Create bar graph
plt.figure(figsize=(9, 5))

plt.bar(
    customer_counts.index,
    customer_counts.values,
    color="green",
    edgecolor="black"
)

# Add title and labels
plt.title("Number of Telecom Customers by Internet Service")
plt.xlabel("Internet Service Type")
plt.ylabel("Number of Customers")

plt.xticks(rotation=30)
plt.tight_layout()

# Display graph
plt.show()

print("Telecom customer bar graph generated successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read telecom customer dataset
df = pd.read_csv("")

# Count customers by Internet Service type
customer_counts = df['InternetService'].value_counts()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    customer_counts,
    labels=customer_counts.index,
    autopct="%1.2f%%",
    startangle=90,
    colors=["orange", "green", "skyblue"],
    explode=(0.05, 0.05, 0.05)
)

plt.title("Share of Telecom Customers by Internet Service")

plt.tight_layout()

# Display pie chart
plt.show()

print("Telecom customer pie chart generated successfully!")
```
